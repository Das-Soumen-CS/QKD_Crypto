```python
import random
import sys

def Bitwise_operator( a,  b):
    # Print bitwise AND operation a & b ; #print("a & b =", bin(a & b)[2:].zfill(5))  #bin() converts the integer to binary
    print("a & b =", format(a & b, '0b'))#bin() converts the integer to binary 
    #Print bitwise OR operation a|b
    print("a | b =", format(a | b, '0b'))
    #print Bitwise XOR operation a ^ b
    print("a ^ b =", format(a ^ b, '0b'))
    # print bitwise left shift operator
    print("a << 1 =", a << 1)
    # print bitwise right shift operator
    print("a >> 1 =", a >> 1)  
    pass


def main():
    stream_len=int(input("Enter the stream length="))
    
    #Take First input
    first_input = random.getrandbits(stream_len)
    print("\n"+str(stream_len)+" bit first_integer_input=",first_input)
    First_binary_input_string = format(first_input, '0b')
    print("First_binary_Input =",First_binary_input_string,"\n")
    
    # Take second Input   
    second_input=random.getrandbits(stream_len)
    print(str(stream_len)+" bit second_integer_input=",second_input)
    Second_binary_input_string = format(second_input, '0b')
    print("second_binary_Input=",Second_binary_input_string,"\n")
    Bitwise_operator(first_input,second_input)
    pass
    
main()
```

```python
from random import randrange

def encryption(plain_text, key):
    cipher_text=plain_text^key
    cipher_text_binary = format(cipher_text, '0b')
    print("cipher_text_binary =",cipher_text_binary)
    return cipher_text
    pass
    
def decryption(cipher_text,key):
    plt_retrieved=cipher_text^key
    plt_retrieved_binary_input_string = format(plt_retrieved, '0b')
    print("plain_text_retrieved_binary =",plt_retrieved_binary_input_string)
    return plt_retrieved
    pass 
    
def key_gen(key_size):
    key=[]
    for i in range(0,key_size):
        key.append(randrange(2))
    print("key is(binary)=",key)
    key=int(''.join(map(str, key)), 2)
    print("key is(decimal)=",key,"\n")
    return key
    pass
    
def main():
    plt_size=int(input("plain text or message size ="))
    key_size=plt_size
   
    #Take plain_text input as binary bits
    plt_input = random.getrandbits(plt_size)
    print("\n"+str(plt_size)+" bit plain_text_integer_input=",plt_input)
    plt_binary_input_string = format(plt_input, '0b')
    print("plain_text_binary =",plt_binary_input_string,"\n")
    
    key=key_gen(key_size)
    cipher_text=encryption(plt_input,key)
    print("The encrypted plain text is = ",cipher_text,"\n")
    plain_text_retrieved=decryption(cipher_text,key)
    print("The decrypted cipher text is = ",plain_text_retrieved)
    pass
    
main()
```
