```python
df=pd.DataFrame(iris_data)
print(df.tail())
```

```python
features = []
for col in df.columns:
    features.append(col)
    #print(col)
print(features)

features_1 = []
for i in range(1,len(features)-1):
    features_1.append(features[i])
print(features_1)

labels = df['Species']
print("class label categorical =\n",labels)

label = pd.Categorical(df['Species']).codes
print("class label neumerical =\n",label)
```

```python
from sklearn.preprocessing import MinMaxScaler
features = MinMaxScaler().fit_transform(df[features_1])
print(features)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#df = pd.DataFrame(df, columns=df.feature_names)
df=df.drop(['Id'], axis=1)
df["class"] = pd.Series(label)
sns.pairplot(df, hue="class", palette="tab10")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from qiskit_algorithms.utils import algorithm_globals

algorithm_globals.random_seed = 123
train_features, test_features, train_labels, test_labels = train_test_split(
    features, label, train_size=0.8, random_state=algorithm_globals.random_seed
)

print("Training data =\n",train_features)
print("type =\n",type(train_features))
print("Training label=\n",train_labels)
```

```python
from sklearn.svm import SVC

svc = SVC()
_ = svc.fit(train_features, train_labels)  # suppress printing the return value
```

```python
train_score_c4 = svc.score(train_features, train_labels)
test_score_c4 = svc.score(test_features, test_labels)

print(f"Classical SVC on the training dataset: {train_score_c4:.2f}")
print(f"Classical SVC on the test dataset:     {test_score_c4:.2f}")
```

```python
from qiskit.circuit.library import ZZFeatureMap
print(features)
num_features = features.shape[1]

feature_map = ZZFeatureMap(feature_dimension=num_features, reps=1)
print(feature_map)
feature_map.decompose().draw(output="mpl", fold=20)
```

```python
from qiskit.circuit.library import RealAmplitudes

ansatz = RealAmplitudes(num_qubits=num_features, reps=3)
print(ansatz)
ansatz.decompose().draw(output="mpl", fold=20)
```

```python
Num_iter=int(input("Please Enter the iteration Number="))
```

```python
from qiskit_algorithms.optimizers import COBYLA ,ADAM

optimizer = COBYLA(maxiter=Num_iter,disp=True, options=None)
print(optimizer)
```

```python
from qiskit.primitives import Sampler
sampler = Sampler()
print(sampler)
```

```python
from matplotlib import pyplot as plt
from IPython.display import clear_output

objective_func_vals = []
plt.rcParams["figure.figsize"] = (12, 6)

def callback_graph(weights, obj_func_eval):
    clear_output(wait=True)
    objective_func_vals.append(obj_func_eval)
    plt.title("Objective function value against iteration")
    plt.xlabel("Iteration")
    plt.ylabel("Objective function value")
    plt.plot(range(len(objective_func_vals)), objective_func_vals)
    plt.show()
```

```python
import time
from qiskit_machine_learning.algorithms.classifiers import VQC

vqc = VQC(
    sampler=sampler,
    feature_map=feature_map,
    ansatz=ansatz,
    optimizer=optimizer,
    callback=callback_graph,
)

# clear objective value history
objective_func_vals = []

start = time.time()
vqc.fit(train_features, train_labels)
elapsed = time.time() - start

print(f"Training time: {round(elapsed)} seconds")
```

```python
train_score_q4 = vqc.score(train_features, train_labels)
test_score_q4 = vqc.score(test_features, test_labels)

print(f"Quantum VQC on the training dataset: {train_score_q4:.2f}")
print(f"Quantum VQC on the test dataset:     {test_score_q4:.2f}")
```
