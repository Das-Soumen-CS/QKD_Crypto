```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

def QBER(N_wrong ):
    Total=int(input("please enter the correct number of recieved bits="))
    N_right=Total-N_wrong
    if(Total>N_wrong):
        QBER_=N_wrong /(N_wrong+N_right)
        print(QBER)
    else:
        print("No of erroneous bits count is greater than Total no of bits transmitted ! iS IMPOSSIBLE")
    pass
  
def N_wrong(n_stray):
    N_dc=float(input("please enter the dark count number ="))
    N_sop=float(input("please enter the ploarisation error number ="))
    N_AP=float(input("please enter the After pulsing number ="))
    N_wrong_count=N_dc+n_stray+N_sop+N_AP
    return N_wrong_count
    
def Genaric_QBER(n_stray,delta_t,Y_0):
    e_0=0.5 #error rate due random background noise
    #Y_0=Back_ground_detection_probability(n_stray,delta_t)
    theta= math.pi / 6
    ER=float(input("Please provide Signals polarization Extinction Ratio="))
    e_pol=(1/ER+1) +(ER-1/ER+1)*(math.sin(theta) ** 2)  #e_pol= probability of incorrect bit recived due to polarization error.
    #n_geo=geometric loss factor , n_atm=atmospheric loss factor ,n_op=no of stray photons per pulse
    # n_point=loss factor due to error in the beam pointing , n_qrx==quantum reciever efficiency
    D_tx=float(input("please provide Transmitter aperature diameter="))
    D_rx=float(input("please provide reciever aperature diameter="))
    D_iv=float(input("please provide estimated divergence of the transmitter telescope="))
    R=float(input("please provide the slant range="))
    n_geo=pow((D_rx/(D_tx+(D_iv*R))),2)
    n_atm=float(input("Atmospheric loss factor ="))
    n_point=float(input("loss factor due to error in the beam pointing ="))
    n_op=float(input("provide no of stray photons per pulse ="))
    n_qrx=float(input("Quantum reciever efficiency through free space to fiber coupling:"))
    t_link=n_geo*n_atm*n_point*n_op*n_qrx
    mu=float(input("please provide signal mean photon number="))
    s=float(input("please provide occurance percentage="))
    exponent=-(t_link*mu*s)
    e=2.718
    temp=1-pow(e,exponent)
    Q_u=Y_0+temp
    print ("gain of signal states =",Q_u)
    E_u=((e_0*Y_0)+(e_pol*temp))/Q_u
    print("\n genaric QBER Value is=",E_u)
    pass
    
def Back_ground_detection_probability(N_stray,delta_t):
    d=float(input("please provide the dark count rate="))
    #delta_t=float(input("please provide the gating window size="))
    #N_stray=float(input("please enter the stray light number ="))
    n_op=float(input("please enter the quantum reciever optical terminals front end optics throughput ="))
    n_qrx=float(input("please provide reciever efficiency in  free space  ="))
    temp_1=4*d 
    temp_1=temp_1*delta_t
    temp_2=N_stray*n_op
    Y_0=temp_1+temp_2 
    print("\n _Back_ground_detection_probability=",Y_0,"\n")
    return Y_0
    pass
    
def N_stray(delta_t):    
    # H_b =brightness of the sky back ground ,FOV =Quantum reciever view of a solid angle ,A_rx=reciver telescope apature area ,B_filter =
    H_b=float(input("Please provide the value of brightness of the sky background")) # either 785 nm or 1550 nm
    FOV=float(input("Please provide the value of quantum reciever field of view in a solid angle"))
    A_rx=float(input("Please provide the value of telescope aperature area"))
    B_filter=float(input("Please provide the value of Reciever filter spectral bandwidth"))
    P_b=(H_b*FOV)+ (A_rx*B_filter)  #P_b= Stray Noise power 
    hv=56 #currently dummy value
    n_stray=(P_b/hv)*delta_t
    return n_stray
    pass
    
def main():
    delta_t=float(input("please provide the gating window size="))
    n_stray=N_stray(delta_t)
    N_wrong_count=N_wrong(n_stray)   
    QBER(N_wrong_count)
    Y_0=Back_ground_detection_probability(n_stray,delta_t)
    Genaric_QBER(n_stray,delta_t,Y_0)
    pass
    
main()
```
