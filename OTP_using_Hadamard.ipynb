```python
# Import all necessary objects and methods for quantum circuits
import matplotlib.pyplot as plt
import qiskit
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit, execute, Aer
from random import randrange
import binascii
# create a quantum circuit for Alice's qubits.
# It's initially set to |0>.
no_qubits=int(input("please provide the no of qubits="))
ascii_after_encryption=[]

def create_circuit(message):
    qreg_SaSg_1 = QuantumRegister(no_qubits) # quantum register with no_qubits=8/16/ qubits
    creg_SaSg_1 = ClassicalRegister(no_qubits) # classical register with no_bits=8/16/8 bits
    mycircuit_SaSg_1 = QuantumCircuit(qreg_SaSg_1,creg_SaSg_1) # quantum circuit with quantum and classical registers
    
    # apply x-gate to change initial states from 0 to 1, preparing message "10001100"
    for m in range(len(message)):
        if message[m]==1:
            mycircuit_SaSg_1.x(qreg_SaSg_1[m])
    # barrier for a better visualization
    mycircuit_SaSg_1.barrier()       
    # apply h-gate to change the basis
    mycircuit_SaSg_1.h(qreg_SaSg_1)
        
    # barrier for a better visualization
    mycircuit_SaSg_1.barrier()

    # create random key and apply x gates and encrypt message
    key=[] 
    for i in range(no_qubits):
        a=randrange(2)
        key.append(a)
        #print(key)
        if a==1:
            mycircuit_SaSg_1 .x(qreg_SaSg_1[i])  
    print("key is=",key)
    mycircuit_SaSg_1 .barrier()
    print(qreg_SaSg_1,creg_SaSg_1)
    mycircuit_SaSg_1.measure(qreg_SaSg_1,creg_SaSg_1)
    print(mycircuit_SaSg_1)
    encryption(mycircuit_SaSg_1,message,key)
    mycircuit_SaSg_1.barrier()
    pass
    
def char_to_ascii(plt_list):
    ascii_list=[]
    for i in plt_list:
        ascii_list.extend(ord(num) for num in i)
    # printing result
    print("The ascii list is := " + str(ascii_list),"\n")
    ascii_to_binary(ascii_list)
    pass

def ascii_to_binary(ascii_list):
    asc_2_bin=[]
    for i in range(len(ascii_list)):
        asc_2_bin.append(bin(ascii_list[i])[2:])   
    print("7 bit_ascii_2_binary =",asc_2_bin)
   
    asc_2_bin_8_bit=[] 
    for i in range(len(asc_2_bin)):
        asc_2_bin_8_bit.append(str(asc_2_bin[i]).rjust(8, '0'))   
    print("8 bit_ascii_2_binary =",asc_2_bin_8_bit,"\n")
    message_split=[]
    # Enter your message as a list of numbers
    for j in range(len(asc_2_bin_8_bit)):
        message = asc_2_bin_8_bit[j]
        #print(len(message))
        for k in range(len(message)):
            #print(message[k])
            message_split.append(int(message[k]))
        #print(message.type())
        message=list(message_split) 
        message_split.clear()
        print("String_",j,"=",message)
        create_circuit(message)
    pass
    
def encryption(mycircuit_SaSg_1,message,key):
    # execute the 
    job = execute(mycircuit_SaSg_1,Aer.get_backend('qasm_simulator'))
    encryption = job.result().get_counts(mycircuit_SaSg_1)
    # display the measurement results with a total count
    print("Encryption", encryption)
    # This converts the measurement result string into a list
    encrypted_message=list(map(int,[*list(encryption.keys())[0]]))

    # We reverse the list since the quantum simulator considers our MSB as LSB
    encrypted_message.reverse()
    print()
    print("Message:",message)
    print("Key:", key)
    print("Encrypted Message in binary:", encrypted_message,"\n") 
    #ascii=""
    temp="".join([str(i) for i in encrypted_message])
    #print(temp)
    ascii=chr(int(temp, 2))
    #print(ascii)
    ascii_after_encryption.append(ascii)
    print("ascii_after_encryption=",ascii_after_encryption)
    pass
    
def main():
    plt=input("Please enter your message: ")
    plt_list=list(plt)  #my plaintext as input string
    print(plt_list)
    char_to_ascii(plt_list)
    pass
    
main()
```
