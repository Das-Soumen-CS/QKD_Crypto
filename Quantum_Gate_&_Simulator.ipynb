```python
from qiskit import QuantumCircuit, Aer
from qiskit.visualization import array_to_latex

#Unitary Simulator
circuit = QuantumCircuit(1) #create circuit of 2 bits
print(circuit.x(0))
display(circuit.draw('mpl'))

unitary_simulator = Aer.get_backend('unitary_simulator')
unitary_simulator_result = unitary_simulator.run(circuit).result()
print(unitary_simulator_result.data())
array_to_latex(unitary_simulator_result.get_unitary())
```

```python
from qiskit import QuantumCircuit, Aer
from qiskit.visualization import array_to_latex 

circuit = QuantumCircuit(2)
circuit.h(0)
circuit.cx(0, 1)
circuit.measure_all()  # <--
display(circuit.draw('mpl'))
print(circuit)

qasm_simulator = Aer.get_backend('qasm_simulator')
qasm_simulator_result = qasm_simulator.run(circuit).result()
qasm_simulator_result.data()
display(qiskit.visualization.plot_histogram(qasm_simulator_result.get_counts()))

memory_result = qasm_simulator.run(circuit, shots=50, memory=True).result()
print(memory_result.get_memory(circuit))


qiskit.visualization.plot_histogram(memory_result.get_counts())
```
